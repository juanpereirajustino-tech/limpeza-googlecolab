# 1. Importação das bibliotecas

```python
import pandas as pd
import numpy as np
```

# 2. Carregamento da base RAIS 2025

```python
from google.colab import files

uploaded = files.upload()

arquivo = list(uploaded.keys())[0]

df = pd.read_excel(arquivo)

df_original = df.copy()

print("Arquivo carregado:", arquivo)
print("Quantidade de linhas:", df.shape[0])
print("Quantidade de colunas:", df.shape[1])
```

# 3. Conhecendo a estrutura da base

```python
print("Dimensões da base:")
print(df.shape)

print("\nNome das colunas:")
print(df.columns.tolist())

print("\nPrimeiras 5 linhas:")
display(df.head())
```

# 4. Tratamento de valores ausentes

```python
nulos = df.isnull().sum()

print("Valores ausentes por coluna:")
print(nulos)

print("\nTotal de valores ausentes na base:", nulos.sum())
```

```python
print("Colunas com valores ausentes:")

colunas_com_nulos = nulos[nulos > 0]

if len(colunas_com_nulos) == 0:
    print("Não existem valores ausentes na base.")
else:
    print(colunas_com_nulos)
```

# 5. Identificação de registros duplicados

```python
duplicadas = df.duplicated(keep=False)

print("Total de registros:", len(df))
print("Registros envolvidos em duplicidades:", duplicadas.sum())
print("Registros únicos:", df.drop_duplicates().shape[0])
print("Registros excedentes caso as duplicidades exatas fossem removidas:",
      len(df) - df.drop_duplicates().shape[0])
```

# 6. Análise das duplicidades

```python
df["Duplicata_Exata"] = df.duplicated(keep=False)

df_duplicatas = df[df["Duplicata_Exata"]].copy()

print("Registros envolvidos em duplicidades exatas:", len(df_duplicatas))
```

```python
df_duplicatas.head(20)
```

```python
contagem_duplicatas = (
    df_duplicatas
    .groupby(
        [col for col in df.columns if col != "Duplicata_Exata"],
        dropna=False
    )
    .size()
    .reset_index(name="Quantidade")
    .sort_values("Quantidade", ascending=False)
)

contagem_duplicatas.head(20)
```

# 7. Análise da RAIS negativa

```python
pd.crosstab(
    df["Ind Rais Negativa"],
    df["Duplicata_Exata"]
)
```

```python
analise_rais_negativa = (
    df.groupby(
        ["Ind Rais Negativa", "Duplicata_Exata"]
    )
    .size()
    .reset_index(name="Quantidade")
)

analise_rais_negativa
```

# 11. Definição dos critérios de limpeza

Foram identificados registros com duplicidade exata considerando todas as variáveis disponíveis na base.

Entretanto, a duplicidade das variáveis disponíveis não permite afirmar que os registros representam o mesmo estabelecimento, pois a base analisada não apresenta um identificador único do estabelecimento.

Dessa forma, as duplicidades exatas foram identificadas e analisadas, mas não serão excluídas automaticamente. Essa decisão evita a remoção indevida de registros que podem representar estabelecimentos distintos com características iguais nas variáveis disponíveis.

Os critérios de limpeza serão aplicados somente aos registros que apresentarem inconsistências identificáveis nos dados.

```python
df_duplicatas.head(20)
```

# 12. Aplicação da limpeza

```python
df_limpo = df.copy()

print("Registros antes da limpeza:", len(df))
print("Registros depois da limpeza:", len(df_limpo))
print("Registros removidos:", len(df) - len(df_limpo))
```

# 13. Validação da base final

```python
print("=== VALIDAÇÃO DA BASE FINAL ===")

print("Registros na base original:", len(df_original))
print("Registros na base final:", len(df_limpo))
print("Registros removidos:", len(df_original) - len(df_limpo))

print("\nValores ausentes na base final:")
print(df_limpo.isnull().sum().sum())

print("\nDuplicidades exatas na base final:")
print(df_limpo.duplicated(keep=False).sum())
```

### Conclusão da validação

A base final permaneceu com 47.339 registros, pois não foram identificados critérios seguros que justificassem a exclusão das duplicidades exatas encontradas.

A análise identificou 7.906 registros envolvidos em duplicidades exatas. Entretanto, a ausência de um identificador único do estabelecimento impede afirmar que esses registros representam necessariamente erros ou repetições indevidas.

Também não foram identificados valores ausentes na base. Dessa forma, a base foi mantida integralmente, preservando os registros para as etapas posteriores de análise.
